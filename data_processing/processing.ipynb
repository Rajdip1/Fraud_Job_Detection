I used ordinal encoding for experience levels due to their inherent hierarchy and one-hot encoding for employment type to avoid imposing artificial order.

Ordinal Encoding

```python
experience_mapping = {
    'Intership' : 0,
    'Entry level' : 1,
    'Associate' : 2,
    'Mid-Senior level' : 3,
    'Director' : 4,
    'Executive' : 5,
    'Not Applicable' : 6,
    'Unknown' : 6
}

df['req_exp_enc'] = df['required_experience'].map(experience_mapping)
```

One hot encoding

```python
df = pd.get_dummies(df, columns=['employment_type'], drop_first=True)
```

```python
print('Before handling: ', df['req_exp_enc'].isna().sum())
df['req_exp_enc'] = df['req_exp_enc'].fillna(6)
print('After handling: ', df['req_exp_enc'].isna().sum())
```

```python
df.drop(columns=['required_experience'], inplace=True)
print("Dropped required_experience column")
```
