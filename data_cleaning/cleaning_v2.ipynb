```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../data/eda_fake_job_postings.csv')
print('Dataset loaded successfully.')
print('shape of the dataset before dropped columns:', df.shape)
df.head()
```

```python
df.info()
print('Found some in job_description missing values.')
```

```python
# Handling missing values in 'job_description' column
df['job_description'] = df['job_description'].fillna("")
print('Missing values in job_description filled with empty strings.')
```

```python
df.info()
print('Missing values in job_description handled successfully.')
```

Conclusion based on EDA, I retained a minimal set of high-signal features including textual job descriptions, key binary indicators, and selected categorical attributes. I also included job description length as a structural feature and excluded high-cardinality or unstable variables to reduce noise and overfitting.

```python
df = df.drop(
    columns=[
        'job_id',
        'location',
        'industry',
        'function',
        'required_education',
        'salary_range',
        'department',
        'jd_char_len'
    ],
    errors='ignore'
)
print('Columns dropped.')
```

```python
print('Shape of the dataset after dropping columns:', df.shape)
```
