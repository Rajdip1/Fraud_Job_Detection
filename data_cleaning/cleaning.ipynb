```python
import pandas as pd
import numpy as np
```

```python
# loading data
df = pd.read_csv('../data/fake_job_postings.csv', index_col=False)
print('Shape of dataset is: ', df.shape)
df.head()
```

Handling columns

```python
# combining text columns into one column
df['job_description'] = (
    df['title'] + " " + df['company_profile'] + " " + df['description'] + " " + df['requirements'] + " " + df['benefits']
)
```

```python
df.head(3)
```

```python
# Categorical and binary columns
categorical_columns = ['location', 'department', 'salary_range', 'employment_type',
               'required_experience', 'required_education', 'industry', 'function']

binary_columns = ['telecommuting', 'has_company_logo', 'has_questions']
```

```python
# Check Missing Values
print("Before handling it: ")
df[['job_description'] + categorical_columns + binary_columns].isna().sum()
```

```python
# Fill NaNs in job_description
df['job_description'] = df['job_description'].fillna("")
```

```python
# Fill NaNs in Categorical Columns
df[categorical_columns] = df[categorical_columns].fillna('Unknown')
```

```python
# just double checking for binary_column
df[binary_columns].isna().sum()
```

```python
print("After handling it: ")
df[['job_description'] + categorical_columns + binary_columns].isna().sum()
```

Droping text columns because i already comibined them in column called "job_description"

```python
df = df.drop(columns=['title', 'company_profile', 'description', 'requirements', 'benefits'], axis='columns')
print('Text Columns are dropped.')
```
